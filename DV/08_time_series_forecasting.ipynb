# DV Assignment 8: Time Series Forecasting

**Objective:** Plot monthly sales and make a simple baseline forecast using a linear trend. This is an introductory forecast, not a guarantee of future sales.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

sales = pd.DataFrame({
    "Order Date": pd.to_datetime([
        "2026-01-05","2026-01-12","2026-02-03","2026-02-19",
        "2026-03-02","2026-03-15","2026-04-04","2026-04-21",
        "2026-05-08","2026-05-20","2026-06-11","2026-06-25"
    ]),
    "Category": ["Technology","Furniture","Office Supplies","Technology",
                "Furniture","Office Supplies","Technology","Furniture",
                "Office Supplies","Technology","Furniture","Office Supplies"],
    "Region": ["East","West","South","East","West","South",
               "East","West","South","East","West","South"],
    "Sales": [1200,850,430,1500,970,510,1800,1100,620,2100,1300,700],
    "Profit": [240,110,80,330,150,95,410,190,125,500,230,140]
})
sales.head()
```

```python
monthly = sales.groupby(sales["Order Date"].dt.to_period("M"))["Sales"].sum()
monthly.index = monthly.index.to_timestamp()
display(monthly)
```

```python
X = np.arange(len(monthly)).reshape(-1, 1)
y = monthly.values
model = LinearRegression()
model.fit(X, y)

future_steps = 3
X_future = np.arange(len(monthly), len(monthly) + future_steps).reshape(-1, 1)
forecast = model.predict(X_future)

future_dates = pd.date_range(monthly.index[-1] + pd.offsets.MonthBegin(1),
                            periods=future_steps, freq="MS")
forecast_series = pd.Series(forecast, index=future_dates)

plt.figure(figsize=(8,4))
plt.plot(monthly.index, monthly.values, marker="o", label="Historical sales")
plt.plot(forecast_series.index, forecast_series.values, marker="o", linestyle="--", label="Baseline forecast")
plt.title("Monthly Sales and 3-Month Baseline Forecast")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.tight_layout()
plt.show()

display(forecast_series.rename("Forecast Sales"))
```
