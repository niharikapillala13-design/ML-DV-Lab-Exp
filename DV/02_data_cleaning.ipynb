# DV Assignment 2: Data Cleaning and Transformation

**Objective:** Handle missing values, rename columns, change data types, and create a calculated field.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    "OrderDate": ["2026-01-01", "2026-01-02", None, "2026-01-04"],
    "Product": ["Laptop", "Chair", "Pen", "Desk"],
    "Sales": [1200, 250, np.nan, 700],
    "Quantity": [2, 3, 10, np.nan]
})
display(df)
```

## Clean missing values and rename columns

```python
df = df.rename(columns={"OrderDate": "Order Date"})
df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")
df["Sales"] = df["Sales"].fillna(df["Sales"].mean())
df["Quantity"] = df["Quantity"].fillna(0)
display(df)
print(df.dtypes)
```

## Create a calculated field

```python
df["Sales per Item"] = df["Sales"] / df["Quantity"].replace(0, np.nan)
display(df)
```
