# DV Assignment 3: Bar and Line Charts

**Objective:** Create category comparisons with a bar chart and time-series trends with a line chart.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

sales = pd.DataFrame({
    "Order Date": pd.to_datetime([
        "2026-01-05","2026-01-12","2026-02-03","2026-02-19",
        "2026-03-02","2026-03-15","2026-04-04","2026-04-21",
        "2026-05-08","2026-05-20","2026-06-11","2026-06-25"
    ]),
    "Category": ["Technology","Furniture","Office Supplies","Technology",
                "Furniture","Office Supplies","Technology","Furniture",
                "Office Supplies","Technology","Furniture","Office Supplies"],
    "Region": ["East","West","South","East","West","South",
               "East","West","South","East","West","South"],
    "Sales": [1200,850,430,1500,970,510,1800,1100,620,2100,1300,700],
    "Profit": [240,110,80,330,150,95,410,190,125,500,230,140]
})
sales.head()
```

```python
monthly_sales = sales.groupby(sales["Order Date"].dt.to_period("M"))["Sales"].sum()
monthly_sales.index = monthly_sales.index.astype(str)

plt.figure(figsize=(8,4))
monthly_sales.plot(kind="bar", title="Monthly Sales", xlabel="Month", ylabel="Sales")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8,4))
monthly_sales.plot(kind="line", marker="o", title="Monthly Sales Trend", xlabel="Month", ylabel="Sales")
plt.tight_layout()
plt.show()
```

```python
region_sales = sales.groupby("Region")["Sales"].sum()
region_sales.plot(kind="bar", title="Sales by Region", xlabel="Region", ylabel="Sales")
plt.tight_layout()
plt.show()
```
