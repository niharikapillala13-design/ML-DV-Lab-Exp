# DV Assignment 7: Heatmaps and Treemaps

**Objective:** Visualize relationships in a correlation heatmap and category proportions in a treemap.

```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

sales = pd.DataFrame({
    "Order Date": pd.to_datetime([
        "2026-01-05","2026-01-12","2026-02-03","2026-02-19",
        "2026-03-02","2026-03-15","2026-04-04","2026-04-21",
        "2026-05-08","2026-05-20","2026-06-11","2026-06-25"
    ]),
    "Category": ["Technology","Furniture","Office Supplies","Technology",
                "Furniture","Office Supplies","Technology","Furniture",
                "Office Supplies","Technology","Furniture","Office Supplies"],
    "Region": ["East","West","South","East","West","South",
               "East","West","South","East","West","South"],
    "Sales": [1200,850,430,1500,970,510,1800,1100,620,2100,1300,700],
    "Profit": [240,110,80,330,150,95,410,190,125,500,230,140]
})
sales.head()
```

```python
numeric = sales[["Sales", "Profit"]]
corr = numeric.corr()

plt.figure(figsize=(5,4))
plt.imshow(corr, interpolation="nearest")
plt.xticks(range(len(corr.columns)), corr.columns)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.colorbar(label="Correlation")
plt.title("Sales and Profit Correlation Heatmap")
for i in range(len(corr)):
    for j in range(len(corr)):
        plt.text(j, i, f"{corr.iloc[i,j]:.2f}", ha="center", va="center")
plt.tight_layout()
plt.show()
```

```python
category_sales = sales.groupby("Category", as_index=False)["Sales"].sum()
fig = px.treemap(category_sales, path=["Category"], values="Sales",
                 title="Treemap: Sales by Category")
fig.show()
```
