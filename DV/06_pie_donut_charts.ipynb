# DV Assignment 6: Pie and Donut Charts

**Objective:** Show how each category contributes to total sales.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

sales = pd.DataFrame({
    "Order Date": pd.to_datetime([
        "2026-01-05","2026-01-12","2026-02-03","2026-02-19",
        "2026-03-02","2026-03-15","2026-04-04","2026-04-21",
        "2026-05-08","2026-05-20","2026-06-11","2026-06-25"
    ]),
    "Category": ["Technology","Furniture","Office Supplies","Technology",
                "Furniture","Office Supplies","Technology","Furniture",
                "Office Supplies","Technology","Furniture","Office Supplies"],
    "Region": ["East","West","South","East","West","South",
               "East","West","South","East","West","South"],
    "Sales": [1200,850,430,1500,970,510,1800,1100,620,2100,1300,700],
    "Profit": [240,110,80,330,150,95,410,190,125,500,230,140]
})
sales.head()
```

```python
category_sales = sales.groupby("Category")["Sales"].sum()

plt.figure(figsize=(6,6))
plt.pie(category_sales, labels=category_sales.index, autopct="%1.1f%%", startangle=90)
plt.title("Sales Share by Category - Pie Chart")
plt.show()
```

```python
plt.figure(figsize=(6,6))
plt.pie(category_sales, labels=category_sales.index, autopct="%1.1f%%",
         startangle=90, wedgeprops={"width": 0.42})
plt.title("Sales Share by Category - Donut Chart")
plt.show()
```
