# DV Assignment 5: Geo Visualization

**Objective:** Show sales across locations on a map. The example uses sample city coordinates; replace them with your dataset's locations.

```python
import pandas as pd
import plotly.express as px
```

```python
geo = pd.DataFrame({
    "City": ["Mumbai", "Delhi", "Bengaluru", "Hyderabad", "Chennai", "Kolkata"],
    "Latitude": [19.0760, 28.6139, 12.9716, 17.3850, 13.0827, 22.5726],
    "Longitude": [72.8777, 77.2090, 77.5946, 78.4867, 80.2707, 88.3639],
    "Sales": [1200, 950, 1100, 800, 700, 650]
})
display(geo)
```

```python
fig = px.scatter_geo(
    geo, lat="Latitude", lon="Longitude", size="Sales", color="Sales",
    hover_name="City", scope="asia", title="Sample Sales by Indian City"
)
fig.show()
```
